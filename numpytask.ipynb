## Task 1

Create this array:

```
10, 20, 30, 40, 50
```

Find:

- Sum
- Average
- Maximum
- Minimum

````python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])

Sum = np.sum(arr1)
Average = np.mean(arr1)
Max = np.max(arr1)
Min = np.min(arr1)

print(f"Sum: {Sum}")
print(f"Average: {Average}")
print(f"Maximum: {Max}")
print(f"Minimum: {Min}")
````

## Task 5

Create this marks array:

```
35, 67, 89, 45, 90, 22, 70, 58
```

Passing marks are `50`.

Find:

- Passed marks
- Failed marks
- Number of passed students
- Number of failed students

````python
marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])

passed = marks[marks >= 50]
failed = marks[marks < 50]

passed_count = len(passed)
failed_count = len(failed)

print("Passed marks:", passed)
print("Failed marks:", failed)
print("Number of passed students:", passed_count)
print("Number of failed students:", failed_count)
````

````python
import numpy as np

# 1. What NumPy is

# NumPy is a Python library used for
# working with arrays and numerical data.

# 2. Creating NumPy arrays

marks = np.array([85, 72, 90, 65, 78])

print("1D Array:")
print(marks)
````

````python
# 3. 1D and 2D arrays

students = np.array([
    [85, 72, 90],
    [65, 78, 88],
    [92, 81, 75]
])

print("\n2D Array:")
print(students)
````

````python
# 4. ndim

print("\nNumber of dimensions:")
print(students.ndim)

# 5. shape

print("\nShape of array:")
print(students.shape)

# 6. size

print("\nTotal number of values:")
print(students.size)
````

````python
# 7. dtype

print("\nData type:")
print(students.dtype)

# 8. Indexing

print("\nFirst mark:")
print(marks[0])

print("\nSecond student first subject mark:")
print(students[1][0])

# 9. Slicing

print("\nFirst three marks:")
print(marks[0:3])

print("\nFirst two students:")
print(students[0:2])
````

````python
# 10. Vectorized operations

print("\nMarks + 5:")
print(marks + 5)
print("\nMarks - 5:")
print(marks - 5)
print("\nMarks * 2:")
print(marks * 2)
print("\nMarks / 2:")
print(marks / 2)
````

````python
# 11. Mathematical functions
print("\nSquare root:")
print(np.sqrt(marks))

print("\nPower:")
print(np.power(marks, 2))

print("\nAbsolute values:")
numbers = np.array([-10, -5, 5, 10])
print(np.abs(numbers))
````

````python
# 12. Sum, mean, min and max
print("\nSum:")
print(np.sum(marks))

print("\nMean:")
print(np.mean(marks))

print("\nMinimum:")
print(np.min(marks))

print("\nMaximum:")
print(np.max(marks))
````

````python
# 13. Conditions
print("\nMarks greater than 80:")
print(marks > 80)
print("\nMarks equal to 90:")
print(marks == 90)

# 14. Filtering
print("\nMarks greater than 80:")
print(marks[marks > 80])
print("\nMarks less than 80:")
print(marks[marks < 80])
````

````python
# 15. Reshape
numbers = np.array([1, 2, 3, 4, 5, 6])
new_array = numbers.reshape(2, 3)
print("\nReshaped Array:")
print(new_array)
````

````python
# 16. Flatten
print("\nOriginal 2D Array:")
print(new_array)
flat_array = new_array.flatten()
print("\nFlattened Array:")
print(flat_array)

# 17. Random numbers
random_numbers = np.random.randint(1, 100, 5)
print("\nRandom Numbers:")
print(random_numbers)
````
