```python
# Task 1
# Create this student DataFrame:
import pandas as pd

data = {"Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
        "Marks": [70,85,60,95,45]}

df = pd.DataFrame(data)

print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
print("Total Students:", len(df["Name"]))
```

```python
# Task 2
greater = df[df.Marks >= 50]
less = df[df.Marks < 50]
print(greater)
print("-----------")
print(less)
```

```python
# Task 3
df["City"] = ["Karachi", "Lahore", "Karachi", "Islamabad","Gujrawala"]
print(df[df.City == "Karachi"])
```

```python
# Task 4
df["Course"] = ['AI','Web','AI','Web','AI']

print(df.groupby("Course")["Marks"].mean())
```

```python
# Task 5
data = {"Name": ["Ali","Sara","Ahmed","Ayesha","Zain"],
        "Marks": [80,None,70,None,90]}

df = pd.DataFrame(data)

print("Missing values count")
print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print("\n",df)
```

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})
```

```python
final_df = pd.concat([df1, df2], ignore_index= True)
print(final_df)
```

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})
```

```python
final_df = pd.merge(students, marks, on= "ID")
print(final_df)
```
